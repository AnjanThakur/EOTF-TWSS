## Recording context
Channels 1–6 were reported as C3, C4, F3, F4, P3, P4. YES/NO answers included speaking. The Chords Web v2.7.0a screenshot confirms Brain mode and a 50 Hz notch selected on all channels. Manufacturer documentation describes Brain as a 45 Hz low-pass. Inspected public v2.7.1a source saves filtered values, but the exact recording build and effective filter response are unverified. Exports are therefore treated as likely already processed; no filters are repeated or undone.
The 500 Hz sampling rate is provisional, inferred from about 60,000 samples in approximately two minutes; the 50 Hz notch is not a sampling rate. Primary plots use sample index and cycles per sample. Mixed responses lack onset markers. See recordings.json for source links and provenance.

## Interpretation
Samples/windows from the same continuous recording are not independent word trials. Recording-level names do not provide within-file task labels. Correlation and amplitude differences cannot identify a neural source, remove speech-muscle contamination, or establish silent-word decoding. Training/testing requires independently labelled sessions and a verified recording protocol.
